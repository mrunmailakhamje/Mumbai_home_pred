```python
#Splitting the data into target and feature variable
x = new_data.drop("price",axis=1)
y = new_data["price"]
```

```python
x.head()      #Features variable
```

```python
y.head()        #Target variable
```

```python
#Splitting data into train and test
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

#Model Selection

```python
#Satistical Model
#1. linear regression

from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)

#prediction
y_pred_lr = lr.predict(x_test)

#r2 score
from sklearn.metrics import r2_score
r2_score_lr = r2_score(y_test,y_pred_lr)

#print r2 score
print("R2 score of linear regression is :",r2_score_lr)
```

```python
#Tree based model
#1. Decision Tree Regressor

from sklearn.tree import DecisionTreeRegressor
dt = DecisionTreeRegressor()
dt.fit(x_train,y_train)

#Prediction
y_pred_dt = dt.predict(x_test)

#r2 score
r2_score_dt = r2_score(y_test,y_pred_dt)

#Print r2 score
print("R2 score of Decision Tree Regressor is :",r2_score_dt)

```

```python
#2. Random Forest
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor()
rf.fit(x_train,y_train)

#prediction
y_pred_rf = rf.predict(x_test)

#r2 score
r2_score_rf = r2_score(y_test,y_pred_rf)

#print r2 score
print("R2 score of Random Forest Regressor is :",r2_score_rf)
```

```python
#3. XG Boast
from xgboost import XGBRegressor
xgb = XGBRegressor()
xgb.fit(x_train,y_train)

#prediction
y_pred_xgb = xgb.predict(x_test)

#r2 score
r2_score_xgb = r2_score(y_test,y_pred_xgb)

#Print r2 score
print("R2 score of XGBoost Regressor is :",r2_score_xgb)
```

```python
#Distance-based Model
#KNN
from sklearn.neighbors import KNeighborsRegressor
knn = KNeighborsRegressor()
knn.fit(x_train,y_train)

#prediction
y_pred_knn = knn.predict(x_test)

#r2 score
r2_score_knn = r2_score(y_test,y_pred_knn)

#Print r2 score
print("R2 score of KNN Regressor is :",r2_score_knn)
```

```python
#MAE and MSE
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error
mae_rf = mean_absolute_error(y_test,y_pred_rf)
mse_rf = mean_squared_error(y_test,y_pred_rf)
rmse_rf =root_mean_squared_error(y_test,y_pred_rf)

print("MAE of Random Forest Regressor is :",mae_rf)
print("MSE of Random Forest Regressor is :",mse_rf)
print("RMSE of Random Forest Regressor is :",rmse_rf)
```

```python
#Features Importance of rf model
feature_imp = pd.Series(rf.feature_importances_,index=x.columns)
feature_imp = feature_imp.sort_values(ascending=False)
sns.barplot(x=feature_imp,y=feature_imp.index)
plt.title("Features Importance of Random Forest Regressor")
plt.xlabel("Feature Importance")
plt.ylabel("Features")
plt.show()
```
